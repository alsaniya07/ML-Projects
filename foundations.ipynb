# Foundations in Machine Learning


## What is Machine Learning?
Teaching computers to learn from data, like humans learn from experience.\
Examples :- Google Translate, Netflix Recommendation Algorithm.\
Instead of programming every step, we give machines **lots of data** and let them figure out\
patterns on their own.



## Traditional Programming vs Machine Learning
Traditional Programming -> Gives rules + data -> get results\
Machine Learning -> Give data + results -> get rules

## AI vs ML vs DL

### 1. Artificial Intelligence (AI)
The broadest concept — any technique that lets machines mimic "intelligent" human behavior 
(reasoning, decision-making, problem-solving), whether it's rule-based or learned.

**Example:** A chess engine like early Deep Blue used hardcoded rules and search trees 
(if opponent moves X, evaluate Y) — no learning involved, but still AI.

---

### 2. Machine Learning (ML)
A subset of AI where the system LEARNS patterns from data instead of being explicitly 
programmed with rules. You give it data + outcomes, it figures out the function.

**Example:** Gmail's spam filter — instead of hardcoding "block if contains word 'lottery'", 
it's trained on thousands of labeled emails (spam/not spam) and learns the patterns itself.

---

### 3. Deep Learning (DL)
A subset of ML that uses **neural networks** with many layers ("deep") to automatically 
learn features from raw data, without manual feature engineering.

**Example:** Face unlock on your phone — a CNN (Convolutional Neural Network) learns 
edges → shapes → facial features → your face, layer by layer, directly from pixel data. 
You didn't tell it "check for nose distance from eyes" — it learned that on its own.

---

### Quick Analogy
- **AI** = the goal (build something smart)
- **ML** = one way to achieve it (learn from data)
- **DL** = a powerful technique within ML (learn using layered neural nets)

### Real-world stack example: ChatGPT
- It's **AI** (broad goal: converse intelligently)
- Built using **ML** (trained on data, not hardcoded rules)
- Specifically uses **DL** (a Transformer — a deep neural network architecture)


## Types of Machine Learning

### 1. Supervised Learning
Model learns from **labeled data** — each input has a known correct output. Goal is to learn 
a mapping function from input → output, then predict on new unseen data.

**Two main types:**
- **Classification** — output is a category/class
  - Example: Email spam detection (spam / not spam), disease diagnosis (positive/negative)
- **Regression** — output is a continuous number
  - Example: Predicting house prices based on sq. footage, location, rooms

**Algorithms:** Linear Regression, Logistic Regression, Decision Trees, Random Forest, SVM, 
k-NN, Neural Networks (when labeled)

**Real-world example:** Netflix predicting your rating for a movie based on your past ratings — 
trained on (user, movie, rating) triples.

### 2. Unsupervised Learning
Model learns from **unlabeled data** — no correct answers given. It finds hidden patterns, 
structures, or groupings on its own.

Only inputs are provided.


**Main types:**
- **Clustering** — group similar data points together
  - Example: Customer segmentation for marketing (grouping users by purchase behavior)
- **Dimensionality Reduction** — compress data while preserving important info
  - Example: PCA used to compress high-dimensional data (e.g. reducing 1000 features to 50) for visualization or faster processing
- **Association Rule Learning** — find relationships between variables
  - Example: Market basket analysis — "customers who buy bread also buy butter" (used in Amazon's "frequently bought together")

**Algorithms:** K-Means, Hierarchical Clustering, DBSCAN, PCA, Apriori

**Real-world example:** Spotify grouping songs/users into clusters to build "Discover Weekly" 
without anyone manually labeling genres.

---

### 3. Semi-Supervised Learning
Uses a **small amount of labeled data + large amount of unlabeled data**. Useful when labeling 
is expensive or time-consuming but you have tons of raw data.

**Real-world example:** Google Photos — you label a few photos with a person's name, and it 
uses that to identify the same person across thousands of unlabeled photos.

---

### 4. Reinforcement Learning (RL)
An **agent** learns by interacting with an **environment**, taking actions, and receiving 
**rewards or penalties**. Goal is to maximize cumulative reward over time — learns through 
trial and error, not from a fixed dataset.

**Key components:** Agent, Environment, State, Action, Reward, Policy

**Real-world example:** AlphaGo/AlphaZero learning to play Go by playing millions of games 
against itself, getting rewarded for winning. Also used in robotics (robot learns to walk) 
and self-driving cars (learning driving policies).

---

### 5. Self-Supervised Learning (worth knowing, especially for LLMs)
A subset of unsupervised learning where the data **labels itself** — part of the input is 
used to predict another part, so no manual labeling needed at all.

**Real-world example:** GPT-style LLMs — trained by predicting the next word in a sentence. 
The "label" (next word) comes from the text itself, not a human annotator.

## Building a machine learning model generally follows these steps:

**1. Define the problem**
Figure out what you're predicting and what type of problem it is — classification, regression, clustering, ranking, etc. Also decide what success looks like (accuracy, F1, RMSE, business metric).

**2. Collect data**
Gather data relevant to the problem — from databases, APIs, scraping, existing datasets, logs, etc.

**3. Explore and clean the data (EDA)**
- Check for missing values, duplicates, outliers
- Understand distributions, correlations, class imbalance
- Visualize relationships between features and target

**4. Feature engineering**
- Create new features from raw data (e.g., extracting day-of-week from a timestamp)
- Encode categorical variables (one-hot, label encoding, embeddings)
- Scale/normalize numerical features
- Handle missing values (imputation)
- Feature selection to remove irrelevant/redundant features

**5. Split the data**
Train/validation/test split (commonly 70/15/15 or 80/10/10), or use k-fold cross-validation. This prevents data leakage and gives an honest estimate of performance.

**6. Choose a model**
Pick an algorithm suited to the problem — linear/logistic regression, decision trees, random forests, gradient boosting (XGBoost/LightGBM), SVM, neural networks, etc. Often start simple as a baseline before going complex.

**7. Train the model**
Fit the model on training data, letting it learn parameters that minimize a loss function.

**8. Tune hyperparameters**
Use grid search, random search, or Bayesian optimization (e.g., Optuna) with cross-validation to find the best hyperparameters.

**9. Evaluate the model**
Test on held-out validation/test data using appropriate metrics:
- Classification: accuracy, precision, recall, F1, ROC-AUC
- Regression: MAE, RMSE, R²

Check for overfitting (great train performance, poor test performance) or underfitting (poor performance everywhere).

**10. Iterate**
Go back — engineer better features, try different models, adjust hyperparameters, get more/better data — based on what the evaluation reveals.

**11. Deploy**
Package the model (e.g., pickle, ONNX, or a REST API via FastAPI/Flask), integrate it into a production system, and serve predictions.

**12. Monitor and maintain**
Track performance over time, watch for data/concept drift, retrain periodically as new data comes in.
