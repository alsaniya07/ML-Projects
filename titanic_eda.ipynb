```python
## Columns to be dropped or fixed
## PassengerId -> Drop Completely, no use
## Name -> No Correlation, completely unique.
## Cabin -> Drop completely, 77% missing and unique strings, so cannot be fixed.
## Ticket -> Completely drop, Unique Ticket Numbers

## Fill NA Age with median.
df['Age'] = df['Age'].fillna(df['Age'].median())
df = df.drop(columns=['PassengerId','Name','Cabin','Ticket'])
df.isnull().sum()
```

```python
rows = df[df['Embarked'].isnull()]
rows
## Remove 61 and 829, doesnt affect the data that much anyways.
# ALTERNATIVE: Modify the existing DataFrame directly without reassignment
df.drop(index=[61, 829], inplace=True)
df.head()
```

```python
## Data Preprocessing - Convert the Data into a model-readable format.

## 1) Seperate Features - Since we are trying to predict who actually survived, split the features
X = df.drop(columns=['Survived'])
Y = df['Survived']
```

```python
## Converting Categorical into Numerical Data
## Perform Simple Label Encoding
## Sex, Male -> 1 , Female -> 0
X['Sex'] =  df['Sex'].map({
    'male' : 1,
    'female' : 0
})
```

```python
## Perform One Hot Encoding
df=pd.get_dummies(X, columns=['Embarked'],drop_first=True)
df
```

```python
## Perform Simple Label Encoding again to get rid of true and false values
X['Embarked_C'] =  df['Embarked_C'].map({
    False : 0,
    True : 1
})

X['Embarked_Q'] =  df['Embarked_Q'].map({
    False : 0,
    True : 1
})

X['Embarked_S'] =  df['Embarked_S'].map({
    False : 0,
    True : 1
})
```

```python
## Feature Scaling - basically scaling all the features to a particular level so that feature columns with a bigger value dont dominate.
## Apply Standardization on Ticket and MinMax on Age
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X[['Fare']] = scaler.fit_transform(X[['Fare']]);
```
